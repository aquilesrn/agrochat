```python
import duckdb, json, pandas as pd, numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from pathlib import Path
import warnings; warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', 40)
pd.set_option('display.float_format', '{:,.3f}'.format)
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams.update({'figure.figsize': (13, 4), 'figure.dpi': 100})

DB_PATH      = '/app/data/duckdb/agrochat.duckdb'
PQ_PATH      = '/app/data/processed/dataset_modelado.parquet'
INF_PATH     = '/app/data/processed/informe_calidad.json'
ENC_PATH     = '/app/data/processed/label_encoders.json'

con = duckdb.connect(DB_PATH, read_only=True)
print('✅ Conexión OK')
print('Tablas:', [r[0] for r in con.execute('SHOW TABLES').fetchall()])
```

## 1. Informe de calidad (generado por validar_datos.py)

```python
if Path(INF_PATH).exists():
    with open(INF_PATH) as f:
        inf = json.load(f)
    df_inf = pd.DataFrame([
        {'métrica': k, 'valor': v}
        for k, v in inf.items()
        if not isinstance(v, (dict, list))
    ])
    display(df_inf.style.set_caption('Informe de calidad de datos').hide(axis='index'))
else:
    print('⚠️  informe_calidad.json no encontrado. Ejecuta validar_datos.py primero.')
```

## 2. Volúmenes por tabla

```python
resumen = []
for tabla in ['avances', 'coyuntura', 'clima_mensual', 'aemet_estaciones']:
    try:
        n = con.execute(f'SELECT COUNT(*) FROM {tabla}').fetchone()[0]
        resumen.append({'tabla': tabla, 'filas': n})
    except:
        resumen.append({'tabla': tabla, 'filas': 'N/A'})

df_res = pd.DataFrame(resumen)

fig, ax = plt.subplots(figsize=(9, 3))
valores = [r for r in df_res['filas'] if isinstance(r, int)]
etiquetas = [df_res['tabla'][i] for i, r in enumerate(df_res['filas']) if isinstance(r, int)]
bars = ax.barh(etiquetas, valores, color=['steelblue','teal','darkorange','gray'])
ax.bar_label(bars, labels=[f'{v:,}' for v in valores], padding=5, fontsize=10)
ax.set_xlabel('Número de registros')
ax.set_title('Volumen por tabla en DuckDB', fontweight='bold')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e3:.0f}k'))
plt.tight_layout(); plt.show()
display(df_res)
```

## 3. Tabla `avances` — cobertura y calidad

```python
# Filas por nivel
df_nivel = con.execute("""
    SELECT nivel,
           COUNT(*) AS filas,
           COUNT(DISTINCT cultivo) AS cultivos,
           COUNT(DISTINCT periodo_anio) AS anios,
           SUM(CASE WHEN outlier_flag THEN 1 ELSE 0 END) AS outliers
    FROM avances GROUP BY nivel ORDER BY filas DESC
""").df()
display(df_nivel.style.set_caption('avances — distribución por nivel').hide(axis='index'))
```

```python
# Nulos en columnas clave
n_total_av = con.execute('SELECT COUNT(*) FROM avances').fetchone()[0]
df_nulos = con.execute("""
    SELECT
        SUM(CASE WHEN sup_avance  IS NULL THEN 1 ELSE 0 END) AS sup_avance_nulos,
        SUM(CASE WHEN prod_avance IS NULL THEN 1 ELSE 0 END) AS prod_avance_nulos,
        SUM(CASE WHEN cod_provincia IS NULL THEN 1 ELSE 0 END) AS cod_provincia_nulos
    FROM avances
""").df().T.rename(columns={0: 'nulos'})
df_nulos['pct'] = (df_nulos['nulos'] / n_total_av * 100).round(2)
display(df_nulos.style.set_caption('avances — nulos en columnas clave'))
```

```python
# Top 12 cultivos por superficie media nacional
df_top = con.execute("""
    SELECT cultivo,
           ROUND(AVG(sup_avance)/1000, 1)  AS sup_media_miles_ha,
           ROUND(AVG(prod_avance), 2)       AS prod_media_miles_t,
           COUNT(DISTINCT periodo_anio)     AS anios_con_datos
    FROM avances
    WHERE nivel='nacional' AND regimen='total'
      AND sup_avance IS NOT NULL AND outlier_flag=FALSE
    GROUP BY cultivo
    HAVING COUNT(*) >= 12
    ORDER BY sup_media_miles_ha DESC
    LIMIT 12
""").df()

fig, ax = plt.subplots(figsize=(13, 4))
ax.bar(df_top['cultivo'], df_top['sup_media_miles_ha'], color='steelblue', alpha=0.85)
ax.set_xticklabels(df_top['cultivo'], rotation=35, ha='right', fontsize=9)
ax.set_ylabel('Superficie media (miles ha)')
ax.set_title('Top 12 cultivos por superficie media — nivel nacional', fontweight='bold')
plt.tight_layout(); plt.show()
display(df_top.style.set_caption('Top 12 cultivos').hide(axis='index'))
```

```python
# Evolución temporal del top 4 cultivos
top4 = df_top['cultivo'].head(4).tolist()
df_evol = con.execute(f"""
    SELECT cultivo, periodo_anio AS anio, periodo_mes AS mes,
           sup_avance
    FROM avances
    WHERE nivel='nacional' AND regimen='total'
      AND cultivo IN ({','.join(["'"+ c +"'" for c in top4])})
      AND sup_avance IS NOT NULL
    ORDER BY cultivo, anio, mes
""").df()
df_evol['periodo'] = df_evol['anio'].astype(str) + '-' + df_evol['mes'].astype(str).str.zfill(2)

fig, axes = plt.subplots(2, 2, figsize=(14, 7))
for ax, cultivo in zip(axes.flatten(), top4):
    d = df_evol[df_evol['cultivo'] == cultivo]
    ax.plot(range(len(d)), d['sup_avance']/1000, linewidth=1.2, color='steelblue')
    ax.set_title(cultivo, fontsize=10, fontweight='bold')
    ax.set_ylabel('Sup. (miles ha)')
    ax.set_xlabel('Periodo (secuencia)')
plt.suptitle('Evolución temporal de superficie — Top 4 cultivos (nacional)', fontweight='bold')
plt.tight_layout(); plt.show()
```

## 4. Tabla `coyuntura` — precios

```python
df_coy_res = con.execute("""
    SELECT seccion,
           COUNT(DISTINCT producto) AS productos,
           COUNT(*) AS registros,
           ROUND(AVG(precio_sem_actual), 2) AS precio_medio,
           ROUND(AVG(ABS(variacion_pct)), 2) AS variacion_media_abs
    FROM coyuntura
    WHERE precio_sem_actual IS NOT NULL
    GROUP BY seccion ORDER BY registros DESC
""").df()
display(df_coy_res.style.set_caption('Resumen por sección de coyuntura').hide(axis='index'))
```

```python
# Top 8 productos con mayor variación media (más volátiles)
df_vol = con.execute("""
    SELECT producto,
           ROUND(AVG(ABS(variacion_pct)), 2) AS volatilidad_media,
           COUNT(*) AS semanas
    FROM coyuntura
    WHERE variacion_pct IS NOT NULL
    GROUP BY producto HAVING COUNT(*) >= 20
    ORDER BY volatilidad_media DESC LIMIT 8
""").df()

fig, ax = plt.subplots(figsize=(12, 3.5))
ax.barh(df_vol['producto'], df_vol['volatilidad_media'], color='darkorange', alpha=0.85)
ax.set_xlabel('Volatilidad media (|variacion_pct|)')
ax.set_title('Top 8 productos más volátiles en precio semanal', fontweight='bold')
plt.tight_layout(); plt.show()
display(df_vol.style.hide(axis='index'))
```

## 5. Tabla `clima_mensual` — cobertura y distribuciones

```python
# Cobertura de estaciones por provincia
df_prov_cl = con.execute("""
    SELECT cod_provincia,
           COUNT(DISTINCT indicativo) AS estaciones,
           COUNT(*) AS registros,
           ROUND(AVG(t_media), 1) AS t_media_hist,
           ROUND(AVG(precip_mm), 1) AS precip_hist_mm
    FROM clima_mensual
    WHERE cod_provincia IS NOT NULL
    GROUP BY cod_provincia ORDER BY estaciones DESC
    LIMIT 15
""").df()
display(df_prov_cl.style.set_caption('Top 15 provincias por nº de estaciones AEMET').hide(axis='index'))
```

```python
# Distribución de temperaturas y precipitación
df_cl = con.execute("""
    SELECT t_media, precip_mm, horas_sol, hr_media
    FROM clima_mensual
    WHERE t_media IS NOT NULL
    USING SAMPLE 30000
""").df()

fig, axes = plt.subplots(1, 4, figsize=(16, 3.5))
vars_plot = [('t_media', 'T. media (°C)', 'teal'),
             ('precip_mm', 'Precipitación (mm)', 'steelblue'),
             ('horas_sol', 'Horas de sol', 'goldenrod'),
             ('hr_media', 'HR media (%)', 'mediumseagreen')]
for ax, (var, label, color) in zip(axes, vars_plot):
    data = df_cl[var].dropna()
    ax.hist(data, bins=35, color=color, edgecolor='white', alpha=0.85)
    ax.set_title(label, fontsize=10, fontweight='bold')
    ax.set_xlabel(label, fontsize=8)
plt.suptitle('Distribución variables climáticas (muestra 30k registros)', fontweight='bold')
plt.tight_layout(); plt.show()
```

```python
# Heatmap cobertura climática: meses con t_media por provincia × año
df_heat = con.execute("""
    SELECT cod_provincia, anio, COUNT(*) AS meses
    FROM clima_mensual
    WHERE cod_provincia IS NOT NULL AND t_media IS NOT NULL
    GROUP BY cod_provincia, anio
""").df()

if not df_heat.empty:
    pivot = df_heat.pivot(index='cod_provincia', columns='anio', values='meses').fillna(0)
    fig, ax = plt.subplots(figsize=(14, max(5, len(pivot)*0.22)))
    sns.heatmap(pivot, cmap='YlGn', ax=ax, linewidths=0.2,
                cbar_kws={'label': 'Meses con t_media'})
    ax.set_title('Cobertura climática — meses con datos por provincia × año', fontweight='bold')
    ax.set_xlabel('Año'); ax.set_ylabel('Código provincia INE')
    plt.tight_layout(); plt.show()
```

## 6. Dataset de modelado (avances × clima)

```python
if Path(PQ_PATH).exists():
    df_mod = pd.read_parquet(PQ_PATH)
    print(f'Filas: {len(df_mod):,}  |  Columnas: {df_mod.shape[1]}')
    print(f'Cultivos: {df_mod["cultivo"].nunique()}  |  Provincias: {df_mod["cod_provincia"].nunique()}')
    display(df_mod.describe().round(3).style.set_caption('Estadísticos del dataset de modelado'))
else:
    print('⚠️ Parquet no encontrado. Ejecuta validar_datos.py primero.')
```

```python
# Distribución del target: rendimiento (miles t / ha)
if Path(PQ_PATH).exists():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))

    df_mod['rendimiento'].plot.hist(bins=60, ax=ax1, color='steelblue',
                                    edgecolor='white', alpha=0.85)
    ax1.set_title('Distribución del target: rendimiento', fontweight='bold')
    ax1.set_xlabel('Rendimiento (miles t/ha)')

    # Rendimiento medio por cultivo (top 15)
    rend_cult = (df_mod.groupby('cultivo')['rendimiento']
                 .mean().nlargest(15).sort_values())
    rend_cult.plot.barh(ax=ax2, color='teal', alpha=0.85)
    ax2.set_title('Rendimiento medio — Top 15 cultivos', fontweight='bold')
    ax2.set_xlabel('Rendimiento medio (miles t/ha)')
    plt.tight_layout(); plt.show()
```

```python
# Mapa de nulos en features (para decidir estrategia de imputación en XGBoost)
if Path(PQ_PATH).exists():
    FEATURES = ['cultivo_enc','provincia_enc','ccaa_enc','anio','mes',
                't_media_c','precip_media_mm','horas_sol_med',
                'hr_media_pct','t_max_abs_c','t_min_abs_c']
    nulos_f = df_mod[FEATURES].isnull().mean().mul(100).round(1).sort_values(ascending=False)
    fig, ax = plt.subplots(figsize=(10, 3))
    nulos_f.plot.bar(ax=ax, color=['red' if v > 20 else 'steelblue' for v in nulos_f])
    ax.set_ylabel('% nulos')
    ax.set_title('Porcentaje de nulos por feature en dataset de modelado', fontweight='bold')
    ax.axhline(20, color='red', linestyle='--', linewidth=1, label='20%')
    ax.legend(); plt.xticks(rotation=35, ha='right'); plt.tight_layout(); plt.show()
    print(nulos_f.to_string())
```
